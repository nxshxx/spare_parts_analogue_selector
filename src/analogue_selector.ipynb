```python
products = pd.read_csv(r"D:\COE PROJECT\spare_parts_analogue_selector\data\products.csv.txt")
failures = pd.read_csv(r"D:\COE PROJECT\spare_parts_analogue_selector\data\failures.csv.txt")
promotions = pd.read_csv(r"D:\COE PROJECT\spare_parts_analogue_selector\data\promotions.csv.txt")
launch_curves = pd.read_csv(r"D:\COE PROJECT\spare_parts_analogue_selector\data\launch_curves.csv.txt")

print(products.columns)
print(failures.columns)
print(promotions.columns)
print(launch_curves.columns)
```

```python
data = products.merge(failures, on="product_id", how="inner")
data = data.merge(promotions, on="product_id", how="inner")
data = data.merge(launch_curves, on="product_id", how="inner")

data.shape
```

```python
data.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

attribute_columns = ["engine_power_kw", "weight_kg", "price_usd"]

scaler = MinMaxScaler()

data[["engine_power_score", "weight_score", "price_score"]] = scaler.fit_transform(data[attribute_columns])

data[["product_id", "engine_power_score", "weight_score", "price_score"]].head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

attribute_matrix = data[["engine_power_score", "weight_score", "price_score"]]

attribute_similarity = cosine_similarity(attribute_matrix)

attribute_similarity[:5, :5]
```

```python
data["category_score"] = data["category"].map(data["category"].value_counts(normalize=True))

data[["product_id", "category", "category_score"]].head()
```

```python
category_similarity = (data["category"].values[:, None] == data["category"].values[None, :]).astype(float)

category_similarity[:5, :5]
```

```python
failure_columns = ["avg_failure_rate", "critical_failures"]

failure_scaler = MinMaxScaler()

data[["failure_rate_score", "critical_failure_score"]] = failure_scaler.fit_transform(data[failure_columns])

data[["product_id", "failure_rate_score", "critical_failure_score"]].head()
```

```python
failure_matrix = data[["failure_rate_score", "critical_failure_score"]]

failure_similarity = cosine_similarity(failure_matrix)

failure_similarity[:5, :5]
```

```python
launch_columns = [
    "month_1_demand",
    "month_2_demand",
    "month_3_demand",
    "month_4_demand",
    "month_5_demand",
    "month_6_demand"
]

launch_scaler = MinMaxScaler()

launch_scores = launch_scaler.fit_transform(data[launch_columns])

launch_similarity = cosine_similarity(launch_scores)

launch_similarity[:5, :5]
```

```python
promotion_similarity = (
    data["promotion_active"].values[:, None] ==
    data["promotion_active"].values[None, :]
).astype(float)

promotion_similarity[:5, :5]
```

```python
overall_similarity = (
    0.30 * attribute_similarity +
    0.25 * launch_similarity +
    0.20 * category_similarity +
    0.15 * failure_similarity +
    0.10 * promotion_similarity
)

overall_similarity[:5, :5]
```

```python
def find_analogue(product_id):
    index = data.index[data["product_id"] == product_id][0]
    scores = overall_similarity[index].copy()
    scores[index] = -1
    best_index = scores.argmax()

    return {
        "target_product": product_id,
        "analogue_product": data.iloc[best_index]["product_id"],
        "similarity_score": round(scores[best_index] * 100, 2)
    }
```

```python
find_analogue("P001")
```

```python
target_index = data.index[data["product_id"] == "P001"][0]
analogue_index = data.index[data["product_id"] == "P021"][0]

print("Product attribute similarity:", round(attribute_similarity[target_index, analogue_index] * 100, 2))
print("Launch curve similarity:", round(launch_similarity[target_index, analogue_index] * 100, 2))
print("Category similarity:", round(category_similarity[target_index, analogue_index] * 100, 2))
print("Failure behaviour similarity:", round(failure_similarity[target_index, analogue_index] * 100, 2))
print("Promotion similarity:", round(promotion_similarity[target_index, analogue_index] * 100, 2))
print("Overall similarity:", round(overall_similarity[target_index, analogue_index] * 100, 2))
```

```python
def get_confidence(similarity_score):
    if similarity_score >= 80:
        return "HIGH"
    elif similarity_score >= 60:
        return "MEDIUM"
    else:
        return "LOW"

get_confidence(99.76)
```

```python
def explain_analogue(product_id):
    index = data.index[data["product_id"] == product_id][0]
    scores = overall_similarity[index].copy()
    scores[index] = -1
    best_index = scores.argmax()

    similarity_score = scores[best_index] * 100
    confidence = get_confidence(similarity_score)

    return {
        "Target Product": product_id,
        "Best Analogue": data.iloc[best_index]["product_id"],
        "Product Attribute Similarity": round(attribute_similarity[index, best_index] * 100, 2),
        "Launch Curve Similarity": round(launch_similarity[index, best_index] * 100, 2),
        "Category Similarity": round(category_similarity[index, best_index] * 100, 2),
        "Failure Behaviour Similarity": round(failure_similarity[index, best_index] * 100, 2),
        "Promotion Similarity": round(promotion_similarity[index, best_index] * 100, 2),
        "Overall Similarity": round(similarity_score, 2),
        "Confidence": confidence
    }
```

```python
test_attributes = data.loc[data["product_id"] == "TEST001", ["engine_power_kw", "weight_kg", "price_usd"]]

test_failure = data.loc[data["product_id"] == "TEST001", ["avg_failure_rate", "critical_failures"]]

test_launch = data.loc[data["product_id"] == "TEST001", launch_columns]

test_attribute_scores = scaler.transform(test_attributes)
test_failure_scores = failure_scaler.transform(test_failure)
test_launch_scores = launch_scaler.transform(test_launch)

print(test_attribute_scores)
print(test_failure_scores)
print(test_launch_scores)
```

```python
test_attribute_similarity = cosine_similarity(test_attribute_scores, scaler.transform(data.iloc[:30][attribute_columns]))[0]

test_failure_similarity = cosine_similarity(test_failure_scores, failure_scaler.transform(data.iloc[:30][failure_columns]))[0]

test_launch_similarity = cosine_similarity(test_launch_scores, launch_scaler.transform(data.iloc[:30][launch_columns]))[0]

test_category_similarity = (data.iloc[:30]["category"].values == "Unknown").astype(float)

test_promotion_similarity = (data.iloc[:30]["promotion_active"].values == "No").astype(float)

test_overall_similarity = (
    0.30 * test_attribute_similarity +
    0.25 * test_launch_similarity +
    0.20 * test_category_similarity +
    0.15 * test_failure_similarity +
    0.10 * test_promotion_similarity
)

best_index = test_overall_similarity.argmax()

print("Best analogue:", data.iloc[best_index]["product_id"])
print("Similarity:", round(test_overall_similarity[best_index] * 100, 2))
```

```python
test_similarity_score = test_overall_similarity[best_index] * 100

if test_similarity_score >= 80:
    test_confidence = "HIGH"
elif test_similarity_score >= 60:
    test_confidence = "MEDIUM"
else:
    test_confidence = "LOW"

print("Target product: TEST001")
print("Best analogue:", data.iloc[best_index]["product_id"])
print("Similarity:", round(test_similarity_score, 2), "%")
print("Confidence:", test_confidence)

if test_confidence == "LOW":
    print("Warning: No strong analogue found. Manual review required.")
```

```python
test_product = data.iloc[0].copy()

test_product["product_id"] = "TEST002"
test_product["category"] = "Excavator"
test_product["engine_power_kw"] = 106
test_product["weight_kg"] = 5120
test_product["price_usd"] = 52000
test_product["avg_failure_rate"] = 0.125
test_product["critical_failures"] = 8
test_product["promotion_active"] = "Yes"
test_product["month_1_demand"] = 21
test_product["month_2_demand"] = 36
test_product["month_3_demand"] = 52
test_product["month_4_demand"] = 67
test_product["month_5_demand"] = 82
test_product["month_6_demand"] = 93

print(test_product)
```

```python
test2_attributes = test_product[attribute_columns].to_frame().T
test2_failure = test_product[failure_columns].to_frame().T
test2_launch = test_product[launch_columns].to_frame().T

test2_attribute_scores = scaler.transform(test2_attributes)
test2_failure_scores = failure_scaler.transform(test2_failure)
test2_launch_scores = launch_scaler.transform(test2_launch)

test2_attribute_similarity = cosine_similarity(
    test2_attribute_scores,
    scaler.transform(data.iloc[:30][attribute_columns])
)[0]

test2_failure_similarity = cosine_similarity(
    test2_failure_scores,
    failure_scaler.transform(data.iloc[:30][failure_columns])
)[0]

test2_launch_similarity = cosine_similarity(
    test2_launch_scores,
    launch_scaler.transform(data.iloc[:30][launch_columns])
)[0]

test2_category_similarity = (
    data.iloc[:30]["category"].values == test_product["category"]
).astype(float)

test2_promotion_similarity = (
    data.iloc[:30]["promotion_active"].values == test_product["promotion_active"]
).astype(float)

test2_overall_similarity = (
    0.30 * test2_attribute_similarity +
    0.25 * test2_launch_similarity +
    0.20 * test2_category_similarity +
    0.15 * test2_failure_similarity +
    0.10 * test2_promotion_similarity
)

top_indices = test2_overall_similarity.argsort()[-5:][::-1]

for index in top_indices:
    print(
        data.iloc[index]["product_id"],
        round(test2_overall_similarity[index] * 100, 2), "%"
    )
```

```python
top_score = test2_overall_similarity[top_indices[0]] * 100
second_score = test2_overall_similarity[top_indices[1]] * 100

difference = abs(top_score - second_score)

print("Best analogue:", data.iloc[top_indices[0]]["product_id"])
print("Best similarity:", round(top_score, 2), "%")
print("Second analogue:", data.iloc[top_indices[1]]["product_id"])
print("Second similarity:", round(second_score, 2), "%")
print("Difference:", round(difference, 2), "%")

if difference <= 1:
    print("Confidence: MEDIUM")
    print("Warning: Multiple near-tied analogues found. Manual review recommended.")
else:
    print("Confidence: HIGH")
```
